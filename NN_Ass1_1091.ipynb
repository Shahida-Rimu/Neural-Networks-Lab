Dataset Creation

```python

# Rule used for labeling:
# If study_hours >= 3 AND attendance >= 70 → Pass (1)
# Otherwise → Fail (0)


data = [
    [2, 60],
    [3, 73],
    [1, 25],
    [4, 80],
    [5, 90],
    [2, 68],
    [3, 75],
    [4, 65],
    [1, 50],
    [5, 85],
    [3, 40],
    [4, 78]
]

labels = [
    0,  # Fail
    1,  # Pass
    0,
    1,
    1,
    0,
    1,
    0,
    0,
    1,
    0,
    1
]
```

Perceptron Initialization

```python

w1 = 0.1   # weight for study hours
w2 = 0.1   # weight for attendance
bias = 0.1

learning_rate = 0.01
```

Activation Function

```python
def activation_function(study_hours, attendance):
    weighted_sum = (w1 * study_hours) + (w2 * attendance) + bias

    # Threshold = 1.0
    if weighted_sum >= 1.0:
        return 1
    else:
        return 0
```

Training Loop

```python
epochs = 20  # number of training cycles

for epoch in range(epochs):
    total_error = 0

    for i in range(len(data)):
        x1 = data[i][0]
        x2 = data[i][1]
        actual = labels[i]

        prediction = activation_function(x1, x2)
        error = actual - prediction
        total_error += abs(error)

        # Weight update rule:
        # new_weight = old_weight + learning_rate * error * input
        w1 = w1 + learning_rate * error * x1
        w2 = w2 + learning_rate * error * x2
        bias = bias + learning_rate * error

    print("Epoch:", epoch + 1, "Total Error:", total_error)
```

User Input Testing

```python
print("\n--- Student Result Prediction ---")

study = float(input("Enter study hours per day: "))
attendance = float(input("Enter attendance percentage: "))

result = activation_function(study, attendance)

if result == 1:
    print("The student is likely to PASS ")
else:
    print("The student is likely to FAIL ")
```
